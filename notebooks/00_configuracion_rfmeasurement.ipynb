```python
from pathlib import Path
import skrf as rf
import rfmeasurement
from rfmeasurement.domain import Measurement, MeasurementContext
from rfmeasurement.validation import validate

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
DATA_ROOT = ROOT / 'data'
CALIBRATION_ROOT = DATA_ROOT / 'calibration'
OUTPUT_ROOT = ROOT / 'results'
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
print('rfmeasurement:', rfmeasurement.__version__)
print('Proyecto:', ROOT)
```

```python
sample = next((CALIBRATION_ROOT / 'open').glob('*.s2p'))
network = rf.Network(sample)
measurement = Measurement(
    data=network,
    context=MeasurementContext(
        dut=f'open standard: {sample.stem}',
        instrument='NanoVNA',
        calibration='uncalibrated/raw',
        frequency_range_hz=(float(network.frequency.start), float(network.frequency.stop)),
    ),
)
report = validate(measurement)
print('Archivo:', sample)
print('Validacion:', 'OK' if not report.has_failures else 'FAIL')
for result in report.results:
    print(result.status.value, result.rule_id, result.explanation or '')
```
